```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("Sales_Dataset (2).csv")
df
```

```python
print(df.describe())    # numerical summary
print(df.describe(include='all'))  # include categorical
```

```python
print(df.isnull().sum())
```

```python
df = df.drop_duplicates()

# fill missing values (example)
df = df.fillna(0)
```

```python
df['Revenue'] = df['Quantity'] * df['Price'] * (1 - df['Discount (%)']/100)

print("Total Revenue:", df['Revenue'].sum())
```

```python
df.groupby('City')['Revenue'].sum().sort_values(ascending=False)
```

```python
df.groupby('Product_Name')['Quantity'].sum().sort_values(ascending=False).head(10)
```

```python
df.groupby('Gender')['Revenue'].sum()
```

```python
import matplotlib.pyplot as plt

df.groupby('Product_Category')['Revenue'].sum().plot(kind='bar')
plt.title("Revenue by Category")
plt.show()
```

```python
df['Payment_Mode'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Payment Mode Distribution")
plt.ylabel('')
plt.show()
```

```python
pd.crosstab(df['Product_Category'], df['Payment_Mode']).plot(kind='bar', stacked=True)
plt.title("Category vs Payment Mode")
plt.show()
```
